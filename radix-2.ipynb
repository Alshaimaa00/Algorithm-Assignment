```python
def countingSort(A, exp):
    n = len(A)
    B = [0] * n
    k = max(A)
    c = [0] * 10
    for i in range(0, n):
        c[int((A[i] / exp) % 10)] += 1
    for i in range(1, 10):
        c[i] += c[i - 1]
    for j in range(n - 1, -1, -1):
        B[c[int((A[j] / exp) % 10)] - 1] = A[j]
        c[int((A[j] / exp) % 10)] -= 1
    for i in range(0, n):
        A[i] = B[i]

def radixSort(A):
    max1 = max(A)
    num_digits = len(str(max1))
    exp = 1
    while max1 // exp > 0:
        countingSort(A, exp)
        exp *= 10

# Testing
A = [34, 7, 23, 32, 5, 62]
print("Before radix sort arr =", A)
radixSort(A)
print("After radix sort arr =", A)
```
